# Program to handle files using exception handling

```python
try:
    with open("filenotfound.txt", "r") as file:
        content=file.read()
except FileNotFoundError:
    print("File not found")
finally:
    print("File program ran completely")
```

### Exception is a base class and it can handle multiple exception in separate except block, allowing for more specific handling of different errors

```python
# Zero division error
try:
    result=100/0
except Exception as e:
    print(f"Error occured as :{e}")
```

```python
try:
    number=int(input("Enter the number"))
    output=100/number
except ValueError:
    print("Not a valid number")
except ZeroDivisionError:
    print("Cannot divide by zero")
else:
    print(f"final output is {output}")
finally:
    print("process is complete..")
```

### Finally block:
- Release resources such as closing files or datababse connection
- perform cleaning of files such as deleting files which are temporary
- Ensure action are occuring even when those exception has arrived.

```python
try:
    file= open("filehandlingexa", "r")
    content=file.read()
    print(content)
except FileNotFoundError:
    print("File not found")
except Exception as e:
    print(f"Unexpected error {e}")
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print('file has been closed')
```
