# Collection Module

```python
data=["HR", "IT", "IT", "IT", "HR", "Construction", "Retail", "Retail", "IT", "IT"]

from collections import Counter, defaultdict
```

```python
count=Counter(data)
count
```

```python
dd=defaultdict(int)
```

```python
for dept in data:
    dd[dept]+=1
print(dict(dd))
```
