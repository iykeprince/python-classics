# Iteration Module
####  itertools: module provided for creating iterators and i will use it for permutations and combinations.

```python
import itertools
```

```python
data=['A','B','C']
print("Permutations")
for perm in itertools.permutations(data):
    print(perm)
```

```python
print('Combinations')
for comb in itertools.combinations(data, 2):
    print(comb)
```
