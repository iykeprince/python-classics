```python
with open("welcome.txt", "r") as source, open("destination.txt", "w") as destinationfile:
    content=source.read()
    destinationfile.write(content)
```

```python
def countfile(filename):
    with open(filename, "r") as file:
        lines=file.readline()
        linecount=len(lines)
        wordcount=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return linecount,wordcount,char_count
```

```python
linecount,wordcount,char_count=countfile("welcome.txt")
print(f"lines: {linecount}, Words: {wordcount}, characters: {char_count}")
```
