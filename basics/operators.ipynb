Comparison Operator

Useful for comparing.

```python
num1=56
num2=60

eqres=num1==num2 # Equality operator
not_eqres=num1 != num2 # Not equal
gt_res=num1>num2 # greater than
lt_res=num1<num2 # less than
gte_res=num1>=num2 # Greater and equal to
lte_res=num1<=num2 # less than and equal

eqres, not_eqres, gt_res, lt_res, gte_res, lte_res
```

### Logical Operators ( for performing boolean logic )

Widely used in conditional statements.

```python
string1=True
string2=False

# Logical operation
and_op = string1 and string2
or_op = string1 or string2
not_op = not string1

and_op, or_op, not_op
```

Identity Operators
`is` and `is not`; Compare the memory location of two objects.

```python
emp1={'name':'Ash', 'empid':4444}
emp2=emp1
emp3={'name':'Nick', 'empid':4433}
```

```python
if emp1 is emp2:
    print('Employee 1 & 2 are same object')
else:
    print('Employee 1 & 2 are two different object entirely.')
```

```python
if emp1 is emp3:
    print('Both objects are the same')
else:
    print('both objects are different')
```
