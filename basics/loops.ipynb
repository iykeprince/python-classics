## Loops in python

Loops are a fundamental programming concept used to execute a block of code repeatedly. Python uses indentation to define the code block that belongs to a loop. There are variables that change with each iteration of the loop.

For loop: Iterate over a sequence.

```python
for i in range(5):
    print(i)
```

## Custom Range and Step size in For Loop

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(10, 0, -1):
    print(i)
```

Iterate over Strings

```python
string_torun="ITOnlineLearning"
for char in string_torun:
    print(char)
```

### While Loop

```python
count=0
while count<5:
    print(count)
    count+=1
```

```python
count=0
while True:
    if count == 100:
        break
    print(count)
    count+=1
```

### Continue

Skips the rest of current iterations and moves to the next one.

```python
for i in range(100):
    if i%2==0:
        continue
    print(i)
```

# Pass Statement

Used when you just want to do nothing in a code. It acts as a placeholder.

```python
for i in range(50):
    if i==3:
        pass
    print(i)
```

### Nested Loops

Loop inside another loop.

```python
for i in range(4):
    for j in range(3):
        print(f"i={i}, j={j}")
```

#### Prime Numbers between 1 and 50

```python
for num in range(2, 51):
    isprime=True
    for i in range(2, num):
        if num%i==0:
            isprime=False
            break
    if isprime:
        print(num)
```
