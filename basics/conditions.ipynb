# Nested Conditions Statements:

Basically, when you have a condition inside another.

````python
# Checking a number whether it is even, odd or negative number

number=int(input('Enter the number:'))
````

````python
if number>=0:
    print('Number is positive')
    if number % 2 == 0:
        print('Number is even')
    else:
        print('Number is odd')
else:
    print('Number is negative')
````

## Exercise

Write a probram that calculates the discount on a purchase based on membership status and the purchase amount applying the best practices we've discussed.

Requirements:
    if the customer is a member:
        - purcase amount >= $100: 20% discount 
        - purchase amount >= $50: 10% discount
        - otherwise: 5% discount.

    if the customer is not a member:
        - purchase amount >= $100: 5% discount.
        - otherwise: No discount

````python
member_status=input("are you a member").strip().lower()
````

````python
try:
    purchaseAmount=float(input("enter the purchase amount"))
except ValueError:
    print("Invalid input")
    exit()

discount=0.0

if member_status =='yes':
    if purchaseAmount>=100:
        discount=0.2
    elif purchaseAmount>=50:
        discount=0.1
    else:
        discount=0.05

elif member_status=='no':
    if purchaseAmount>=100:
        discount=0.05

else:
    print("Invalid membership entered")
    exit()

discountAmount=purchaseAmount*discount
finalprice=purchaseAmount-discountAmount

print(f"Purchase amount: ${purchaseAmount:.2f}")
print(f"Discount applied: ${discount*100}%")
print(f"Discount amount: ${discountAmount:.2f}")
print(f"Final price: ${finalprice:.2f}")

````
