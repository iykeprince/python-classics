### Mathematical Operations in Sets
- Union: Combined elements from both sets ensuring that there are no duplicates exist in your result

```python
numbers1={56,57,58,59,34,8,9}
numbers2={64,66,56,59,33,7,2}

```

```python
union_numbers=numbers1.union(numbers2)
union_numbers
```

```python
len(union_numbers)
```

### Intersection: Elements common to both sets

```python
inter_numbers=numbers1.intersection(numbers2)
inter_numbers
```

## Difference: Element exists in set1 but not in set2

```python
differences_out=numbers1.difference(numbers2)
differences_out
```

### Symmetric Difference: Elements unique to each set and it will include all the elements except the common ones(exclude commong elements)

```python
sym_out=numbers1.symmetric_difference(numbers2)
sym_out
```

```python
len(sym_out)
```

### Exercise1: Two sets of friends and you need to find out the common friends (mutual friends)

```python
def mutual_list(friends_a, friends_b):
    return friends_a.intersection(friends_b)

rakesh_friends={'Ashi', 'David', 'Damia', 'Ciaro', "Chris"}
rambi_friends={'David', 'Damia', 'Lieao', "Chris", "Tsu"}

mutual_friends=mutual_list(rakesh_friends, rambi_friends)
mutual_friends
```
