## Dictionary Comprehension

```python
transactions={
    "CVX33533": {"amount": 3300, "status":"completed"},
    "CVX33534": {"amount": 2430, "status":"pending"},
    "CVX33535": {"amount": 1450, "status":"completed"},
    "CVX33536": {"amount": 2900, "status":"failed"},
}
```

```python
completed_transactions={txid: details for txid, details in transactions.items() if details["status"] == "completed"}
completed_transactions
```

```python
pending={txid: details for txid, details in transactions.items() if details["status"] == "pending"}
pending
```

```python
# Real questions: Count frequency of elements in a list
```

```python
numbers=[1,2,3,2,2,3,4,4,5,6]
frequency={}
for num in numbers:
    frequency[num]=frequency.get(num,0)+1

print(frequency)
```

## Common Errors in Dictionary      

```python
try:
    item=transactions["Item"]
except KeyError:
    print('Discount not found')
    item=0

print(item)
```

```python
from collections import defaultdict
trncount=defaultdict(int)
```

```python
transactions=[
    {"id":"CDFSF56", "tstatus":"Pending"},
    {"id":"SDF3233", "tstatus": "Completed"},
    {"id":"CXD4533", "tstatus": "Pending"},
    {"id":"CXD4533"}
]
```

```python
for tx in transactions:
    status=tx.get("tstatus", "unknown")
    trncount[status]+=1
print(trncount)
```
