Merge: used often with + or extend() ,while duplicating listing

```python
list1=[12,34,33,45,63,89]
list2=[33,44,65,90]

new_list_formed=list1+list2
duplicate_list_formed=new_list_formed
new_list_formed
```

```python
# set() is used to remove duplicate element. it doesn't retain order

removed_dup=list(set(duplicate_list_formed))
removed_dup
```

```python
# retain order in list using for loop
unique_elem_list=[]
for item in new_list_formed:
    if item not in unique_elem_list:
        unique_elem_list.append(item)

print(unique_elem_list)
```

any or all: checks conditions on list

any: check if any element satisfy the conditions in list
all: check if all element satisfy the condition

```python
nums=[21,41,51]

has_any_even=any(x%2==0 for x in nums)
has_any_even
```

```python
all_even=all( x > 20 for x in nums)
all_even
```

# Exercise 2
List of products, filter product based on category user has selected

```python
products=[
    {"product_name": "PA", "price": 3443, "categ": "Electronics"},
    {"product_name": "PB", "price": 3432, "categ": "Clothing"},
    {"product_name": "PC", "price": 3443, "categ": "Electronics"}    
]
```

```python
def filter_prod(products, cat):
    return [product for product in products if product.get('categ') == cat]
```

```python
cat_selected=filter_prod(products, "Clothing")
cat_selected
```
