Python *args & *kwargs

```python
def calculate_sum(*args):
    return sum(args)
```

```python
calculate_sum(34,342,4)
```

```python
def print_bookingdetails(**kwargs):
    print(type(kwargs))
    for key,value in kwargs.items():
        print(f"{key.capitalize()}:{value}")
```

```python
print_bookingdetails(name="Ash", age=32, address="123 High St")
```

## Return & Pass

##### return statements allows you to return value to the caller. Functions can return any data type and even multiple values.

```python
def cal_rect_area_and_para(len, wid):
    par=2*(len+wid)
    area=len*wid
    return area,par
```

```python
area,par=cal_rect_area_and_para(5,10)
print(f"Area is {area}")
print(f"Perimeter is {par}")
```

### Examples & Exercises

```python
def reportEmployee(name, *desg, **avgsalary):
    print(f"Employee who created the report: {name}")
    print("Designations:", ", ".join(desg))
    print("avg salary:")
    for desg,avg_salary in avgsalary.items():
        print(f"{desg.capitalize()}:{avg_salary}")
```

```python
reportEmployee("Ashi", "IT", "HR", "FINANCE", IT=56999, HR=588999, FINANCE=3902933)
```
