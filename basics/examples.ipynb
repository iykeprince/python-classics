## Health Domain
### Example 3: Categorize patients based on BMI. Function and map, we can classify them into groups like underweight, normal, overweight and obese

- bmi < 18.5 --- underweight
-  18.5 <= bmi <= 24.0 -- Normal
-  25 <= bmi < 29 ---- overwieght
- else Obese

```python
def category_bmi(bmi):
    if bmi < 18.5:
        return "Underweight"
    elif 18.5 <= bmi < 24.0:
        return "Normal"
    elif 25 <= bmi < 29:
        return "Overweight"
    else:
        return "Obese"
```

```python
bmi_list = [15, 17, 22, 26, 17, 12, 34]
```

```python
categorized_bmi=list(map(category_bmi, bmi_list))
```

```python
categorized_bmi
```

## Filter patients with high risk. Age above 60 or having more than 2 comorbidities. Filter we extract this subset

```python
patients=[
    {'name':'Mike', 'age':30,'comborbidities':1},
    {'name':'Tomas', 'age':70,'comborbidities':3},
    {'name':'Max', 'age':50,'comborbidities':2},
    {'name':'Tauren', 'age':40,'comborbidities':3},
]
```

```python
patients
```

```python
# use filter to identify high risk patients
high_riskpatients=list(filter(lambda patient:patient["age"]>60 or patient['comborbidities']>2, patients))
```

```python
high_riskpatients
```

#### Log level information usually in the code for developers to access

```python
log_levels=["error","warn","info","error","debug","warn",None,"info"]
```

## Retail Domain
### EXAMPLE 4: 
#### Calculate total cost of item added in your cart

```python
def calculate_totalitems_incart(price, quantity):
    return sum(price*quantity for price, quantity in zip(price,quantity))
```

```python
totalitempurchased=calculate_totalitems_incart((23,15,100,300),(1,2,3,4))
```
