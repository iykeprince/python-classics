## Inheritance

Why use Inheritance
- It makes it easy to extend code in the future
- It gives meaningful relationship between classes
- Maintainability
- Reusability

````python
# Single Inheritance
````

````python
# parent class
class Employee:
    def __init__(self, name, empid, empdept):
        self.name=name
        self.empid=empid
        self.empdept=empdept
    
    def display_details(self):
        print(f" Name : {self.name} , ID : {self.empid} , Dept: {self.empdept}")
````

````python
#child class (inherits from parent class)
class Manager(Employee):
    def __init__(self, name, empid, empdept, team_size):
        super().__init__(name, empid, empdept)
        self.team_size=team_size

    def showcase_team(self):
        print(f"Manager {self.name} is holding a size of {self.team_size}")
````

````python
manager=Manager('Ash', 'EMP_45345343', "Engineering solution", 54)
manager.display_details()
manager.showcase_team()
````

## Multi-Parent Inheritance

````python
# SMart device app - Phone nad Camera

class Phone:
    def __init__(self, phone_number):
        self.phone_number=phone_number
    def give_calls(self):
        print(f"Calling from {self.phone_number}")
````

````python
class Camera:
    def __init__(self, res):
        self.res=res
    def take_selfies(self):
        print(f"taking selfies oes size {self.res}")
````

````python
class SmartApp(Phone, Camera):
    def __init__(self, phone_number, res, brand):
        Phone.__init__(self,phone_number)
        Camera.__init__(self, res)
        self.brand=brand

    def device_info(self):
        print(f"This is {self.brand} smart app")
````

````python
obj1=SmartApp("78789876","108MP","SONY")
````

````python
obj1.give_calls()
````

````python
obj1.take_selfies()
````

````python
obj1.device_info()
````

## Multilevel Inheritance

````python
class Person:
    def __init__(self, name, id):
        self.name=name
        self.id=id
    def get_details(self):
        return f"Name: {self.name} , ID: {self.id}"
````

````python
class Student(Person):
    def __init__(self, name, id, grade):
        super().__init__(name, id)
        self.grade=grade
    def get_details(self):
        return f"Name: {self.name}, ID: {self.id}, Grade: {self.grade}"
````

````python
class GradStudent(Student):
    def __init__(self, name, id, grade,thesis):
        super().__init__(name, id, grade)
        self.thesis=thesis
    def get_details(self):
        return f"Name: {self.name}, ID: {self.id}, Grade: {self.grade}, Thesis:{self.thesis}"
````

````python
obj2=GradStudent("Dave","6736778", "A", "AI Detection for diseases")
obj2.get_details()
````

````python
class PaymentWays:
    def __init__(self, accountholder):
        self.accountholder=accountholder
    
    def pay(self,amount):
        print(f"{self.accountholder} is paying {amount} using generic payment method")
````

````python
#Credit Card
class CreditCard(PaymentWays):
    def __init__(self, accountholder,card_number):
        super().__init__(accountholder)
        self.card_number=card_number
    
    def pay(self, amount):
        print(f"{self.accountholder} is paying {amount} using Credit Card with {self.card_number[-3:]}")
````

````python
#Paypal Account
class Paypal(PaymentWays):
    def __init__(self, accountholder, emailId):
        super().__init__(accountholder)
        self.emailId=emailId


    def pay(self, amount):
        print(f"{self.accountholder} is paying {amount} using Paypal Account with {self.emailId}")
````

````python
class BankTransfer(PaymentWays):
    def __init__(self, accountholder, bankaccount):
        super().__init__(accountholder)
        self.bankaccount=bankaccount

    def pay(self, amount):
        print(f"{self.accountholder} is paying {amount} through bank account {self.bankaccount}")
````

````python
payments=[
    CreditCard("Sarah", "3234243433"),
    Paypal("David", "david@awol.com"),
    BankTransfer("Walter", "HSBC")
]
````

````python
for method in payments:
    print(f"Processing payment for {method.accountholder} ({type(method).__name__})")
    method.pay(100)
````

### Interfaces(abstract base classes)

````python
from abc import ABC, abstractmethod
class Device(ABC):
    @abstractmethod
    def power_on(self):
        pass
    
````

````python
class Laptop(Device):
    def power_on(self):
        return "Laptop is turning on"
````

````python
class Printer(Device):
    def power_on(self):
        return "Printer is turning on"
````

````python
def start_device(dev):
    print(dev.power_on())

lap=Laptop()
printer=Printer()
````

````python
start_device(lap)
````

````python
start_device(printer)
````

# Abstraction

````python
# Abstract = show what, hide how.
````

### Why Abstraction?
- Reduces complexity for users of your code
- Keeps codebases modular maintainable
- Encouragees reuse via common interfaces

Data Abstraction
- Hiding the actual representation of the data and exposing only the operations that can be performed on it.

```python
class BankAccount 
    def __init__(self,balance):
        self.__balance=balance #hidden(name mangled)
```

Process Abstraction
- Hiding the detail of how a process or operation is carried out, exposing only what the process does.


````python
from abc import ABC, abstractmethod
class dem(ABC):
    @abstractmethod
    def method1(self):
        print("Abstract method")
        return 
    
    def method2(self):
        print("Concrete method")
        
````

````python
obj=dem()
````

````python
class Concreteclass(dem):
    def method1(self):
        super().method1()
    def method2(self):
        super().method2()
````

````python
obj=Concreteclass()
obj.method1()
````

````python
obj.method2()
````

````python
from abc import ABC, abstractmethod
class PaymentGateway(ABC):
    def name(self):
        print(f"Using payment gatways : {self.__class__.__name__}")

    @abstractmethod
    def authorize(self,amount: float) -> str:
        """return authorization ID"""
        pass

    @abstractmethod
    def capture(self,authid: str) -> None:
        """return authorization ID"""
        pass
````

````python
#Concreate Implementation
class StripeGateway(PaymentGateway):
    def authorize(self, amount:float) -> str:
        print(f"Stripe authorizing ${amount}")
        return "authstripe_123"
    
    def capture(self,authid: str) -> None:
        print(f"Stripe Capturing funds for {authid}")
````

````python
#Concreate Implementation
class PaypalGateway(PaymentGateway):
    def authorize(self, amount:float) -> str:
        print(f"Paypal authorizing ${amount}")
        return "paypal_123"
    
    def capture(self,authid: str) -> None:
        print(f"Paypal Capturing funds for {authid}")
````
