```python
gasoline={
    'gas',
    'E85 / Unleaded',
    'Unleaded',
    'Unleaded; Unleaded / E85', 
    'Unleaded / E85',
    'E85 / Unleaded; Unleaded', 
    'Premimum Unleaded / Unleaded',
    'E85 / Premium Unleaded; E85 / Unleaded',
    'E85', 'E85 / Premium Unleaded',
    'Compressed Natural Gas / Lp',
    'Compressed Natural Gas / Unleaded',
    'Compressed Natural Gas',
    'E85 / Unleaded; Unleaded / Unleaded',
    'Diesel / Premium Unleaded', 
    'E85 / Unleaded; Unleaded / E85', 
    'Unleaded / Unleaded', 
    'Diesel; Unleaded',
    'Unleaded / Premium Unleaded',
    'Premium Unleaded / E85',
    'M85 / Unleaded',
    'E85',
    'Premium Unleaded / Unleaded',
    'Premium Unleaded',
    'Compressed Natural Gas; Unleaded',
    'Premium Unleaded; Unleaded',
}
biodiesel={
    'Biodiesel'
}
hybrid={
    'Electric / Unleaded',
    'Electric',
    'Electric / E85',
    'Electric / Premium Unleaded',
    'Electric / E85',
    'Electric / Hydrogen',
    'Electric / Premium Unleaded; Electric / Unleaded',
    'Electric / Premium Unleaded; Premium Unleaded',
    'Unleaded / Electric'
}

other={
    'Hydrogen',
    'Premium Unleaded / Natural Gas',
    'Compressed Natural Gas / Lpg',
    'Compressed Natural Gas', 'Propane',
    'Flex Fuel Vehicle'

}

def preprocess_fuel_type(x):
    if x in gasoline:
        return "gasoline"
    elif x in biodiesel:
        return "biodiesel"
    elif x in hybrid:
        return "hybrid"
    else:
        return "other"
    
df['fuel_type']=df['fuel_type'].apply(preprocess_fuel_type)
```
